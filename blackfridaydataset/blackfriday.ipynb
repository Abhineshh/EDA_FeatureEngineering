## Problem Statement

- A retail company 'ABC Private Limited" wants to understand the customer purchase behaviour(specifically purchase amt) against various products of different categories. They have shared purchase summary of various customers for selected high volume products from last month. The dataset also contins customer demographics (age, gender, marital status, city_type, stay_in_current_city), product details(product_id and product category) and total purchase_amount from last month.

```python
# create numerical columns to represent the city_category column 
# and delete city_category column

cityCatdf = pd.get_dummies(df["City_Category"], drop_first=True)
df = pd.concat([df, cityCatdf],axis = 1).drop('City_Category', axis = 1)
```

# Handling Missing Values

```python
df.isna().sum()
```

since our data is train + test , purchase is supposed to have null values
we need to handle only product_category_2 and 3

```python
df["Product_Category_2"].value_counts()
```

- we will take 'mode' of the column and fill it in the null cells. we take mode because mode is nothing but value that occur frequently. 
- the reason we don't take mean is that mean creates a totally new values unrelated to the data we have.

```python
mode_val = df["Product_Category_2"].mode()[0]
# here we are getting a Pandas Series so we take only the value
mode_val
```

```python
df["Product_Category_2"] = df["Product_Category_2"].fillna(mode_val)
```

```python
df["Product_Category_2"].isna().sum() # checking if null
```

```python
mode_val = df["Product_Category_3"].mode()[0]
df["Product_Category_3"] = df["Product_Category_3"].fillna(mode_val)
df["Product_Category_3"].isna().sum() 
```

```python
print(df["Product_Category_2"].value_counts())
print(df["Product_Category_3"].value_counts())
print(df["Product_Category_2"].isna().sum())
print(df["Product_Category_2"].isna().sum())
```

```python
# handing Stay_In_Current_City_Years columns values
df['Stay_In_Current_City_Years'].unique()
# here in this column we basically have numbericall values but in string format 
# with '4+" as one of the values which is not numerically acceptable
#so we convert it to 4 and other to numbers aswell
```
